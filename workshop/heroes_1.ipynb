```python
import sqlalchemy as sa



```

```python
engine = sa.create_engine('sqlite:///../heroes_sql/heroes.db')
connection = engine.connect()
```

```python
query = sa.text('SELECT * FROM users')
data_cursor = connection.execute(query)
data = data_cursor.fetchall()
print(data)
```

```python
metadata = sa.MetaData()

heroes_table_1 = sa.Table(
    'heroes',
    metadata,
    sa.Column('hero_id', sa.Integer, primary_key=True, nullable=False),
    sa.Column('name', sa.String, nullable=False, unique=True)
)
```

```python
query_ = sa.select(heroes_table_1)
data_cursor_ = connection.execute(query)
heroes = data_cursor_.fetchall()
print(heroes)
```

```python
q = sa.select(heroes_table_1).limit(5)
dc = connection.execute(q)
d = dc.fetchall()
print(d)
```

```python
q = sa.select(heroes_table_1).offset(4).limit(6)
dc = connection.execute(q)
d = dc.fetchall()
print(d)
```

```python
q = sa.select(heroes_table_1).order_by(heroes_table_1.c.name).limit(5)
dc = connection.execute(q)
d = dc.fetchall()
print(d)
```

```python
q = sa.select(heroes_table_1).order_by(sa.desc(heroes_table_1.c.name)).limit(5)
dc = connection.execute(q)
d = dc.fetchall()
print(d)
```

```python
q = sa.select(heroes_table_1).where(heroes_table_1.c.hero_id > 5 ).limit(5)
dc = connection.execute(q)
d = dc.fetchall()
print(d)
```
